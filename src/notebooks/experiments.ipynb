# Lab 02: Point Operations & Linear Filters Experiments

```python
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import cv2

# Đưa thư mục gốc dự án vào sys.path
ROOT_DIR = Path().resolve().parent if Path().resolve().name in ["notebooks", "src"] else Path().resolve()
if str(ROOT_DIR) not in sys.path:
    sys.path.append(str(ROOT_DIR))

from src.io.image_io import read_image
from src.point_operations.negative import negative_image
from src.point_operations.brightness import change_brightness
from src.point_operations.contrast import change_contrast
from src.point_operations.threshold import threshold_image
from src.linear_filters.gaussian_filter import gaussian_filter
from src.linear_filters.mean_filter import mean_filter
from src.linear_filters.sharpening import sharpening as sharpen_image

print("Import thành công!")
```

```python
# Load và hiển thị ảnh gốc
input_path = ROOT_DIR / "data" / "input" / "text.jpg"
img_bgr = read_image(input_path)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# 1. Point Operations
img_neg = cv2.cvtColor(negative_image(img_bgr), cv2.COLOR_BGR2RGB)
img_bright = cv2.cvtColor(change_brightness(img_bgr, beta=50), cv2.COLOR_BGR2RGB)
img_contrast = cv2.cvtColor(change_contrast(img_bgr, alpha=1.5), cv2.COLOR_BGR2RGB)
img_thresh = threshold_image(img_bgr, threshold=128)

fig, axes = plt.subplots(2, 2, figsize=(10, 10))
axes[0, 0].imshow(img_neg)
axes[0, 0].set_title("Negative")

axes[0, 1].imshow(img_bright)
axes[0, 1].set_title("Brightness (+50)")

axes[1, 0].imshow(img_contrast)
axes[1, 0].set_title("Contrast (1.5)")

axes[1, 1].imshow(img_thresh, cmap="gray")
axes[1, 1].set_title("Threshold (128)")

for ax in axes.ravel():
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
# 2. Linear Filters
img_gauss = cv2.cvtColor(gaussian_filter(img_bgr, kernel_size=5, sigma=1.0), cv2.COLOR_BGR2RGB)
img_mean = cv2.cvtColor(mean_filter(img_bgr, kernel_size=5), cv2.COLOR_BGR2RGB)
img_sharp = cv2.cvtColor(sharpen_image(img_bgr), cv2.COLOR_BGR2RGB)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img_gauss)
axes[0].set_title("Gaussian Blur")

axes[1].imshow(img_mean)
axes[1].set_title("Mean Blur")

axes[2].imshow(img_sharp)
axes[2].set_title("Sharpening")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```
